**Loading the selected files**

```python
import pandas as pd
import os

data_path = "/content/financial_market_data"

selected_files = {
    "NIFTY 50": "NIFTY 50_Data.csv",
    "NIFTY Bank": "NIFTY BANK_Data.csv",
    "NIFTY IT": "NIFTY IT_Data.csv",
    "NIFTY Auto": "NIFTY AUTO_Data.csv",
    "NIFTY Pharma": "NIFTY PHARMA_Data.csv",
    "NIFTY FMCG": "NIFTY FMCG_Data.csv",
    "NIFTY Energy": "NIFTY ENERGY_Data.csv",
    "NIFTY Metal": "NIFTY METAL_Data.csv",
    "NIFTY Midcap 100": "NIFTY MIDCAP 100_Data.csv",
    "NIFTY Next 50": "NIFTY NEXT 50_Data.csv"
}

market_data = {}

for index_name, filename in selected_files.items():
    file_path = os.path.join(data_path, filename)

    df_temp = pd.read_csv(file_path)

    # Convert Date column to datetime
    df_temp["Date"] = pd.to_datetime(df_temp["Date"])

    # Sort by date
    df_temp = df_temp.sort_values("Date").reset_index(drop=True)

    market_data[index_name] = df_temp

    print(f"{index_name}: {df_temp.shape}")
```

**Cleaning and validating the market data**

```python
clean_market_data = {}

for index_name, df_temp in market_data.items():

    df_clean = df_temp.copy()

    # Convert Date to datetime
    df_clean["Date"] = pd.to_datetime(df_clean["Date"], errors="coerce")

    # Convert OHLC columns to numeric
    numeric_columns = ["Open", "High", "Low", "Close"]

    for col in numeric_columns:
        df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

    # Remove rows where Date or Close is missing
    df_clean = df_clean.dropna(subset=["Date", "Close"])

    # Remove duplicate dates
    df_clean = df_clean.drop_duplicates(subset=["Date"])

    # Sort by date
    df_clean = df_clean.sort_values("Date").reset_index(drop=True)

    clean_market_data[index_name] = df_clean

    print(f"{index_name}: {len(df_clean)} rows after cleaning")
```

**Creating the Master Dataset**

```python
master_data = []

for index_name, df_clean in clean_market_data.items():

    # Create a copy
    temp = df_clean.copy()

    # Add index name
    temp["Index"] = index_name

    # Keep required columns
    temp = temp[
        ["Date", "Index", "Open", "High", "Low", "Close"]
    ]

    master_data.append(temp)

# Combine all indices
master_df = pd.concat(master_data, ignore_index=True)

# Sort by date and index
master_df = master_df.sort_values(
    ["Date", "Index"]
).reset_index(drop=True)

print("Master dataset created successfully!")

print("Rows:", master_df.shape[0])
print("Columns:", master_df.shape[1])

display(master_df.head(10))
```

```python
def calculate_max_drawdown(data):

    prices = data["Close"]

    running_peak = prices.cummax()

    drawdown = (
        (prices - running_peak) /
        running_peak
    ) * 100

    return drawdown.min()


drawdown_results = []

for index_name, data in master_df.groupby("Index"):

    max_drawdown = calculate_max_drawdown(data)

    drawdown_results.append({
        "Index": index_name,
        "Maximum_Drawdown_%": max_drawdown
    })

drawdown_summary = pd.DataFrame(drawdown_results)

drawdown_summary["Maximum_Drawdown_%"] = (
    drawdown_summary["Maximum_Drawdown_%"].round(2)
)

display(drawdown_summary)
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    drawdown_summary["Index"],
    drawdown_summary["Maximum_Drawdown_%"]
)

plt.title("Maximum Historical Drawdown")
plt.xlabel("Market Index")
plt.ylabel("Maximum Drawdown (%)")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Keep original master dataset unchanged
analysis_df = master_df.copy()

# Identify the known NIFTY IT discontinuity
problem_date = pd.Timestamp("2004-05-28")

problem_mask = (
    (analysis_df["Index"] == "NIFTY IT") &
    (analysis_df["Date"] == problem_date)
)

print("Problematic observation found:",
      problem_mask.sum())

# Exclude this observation from return/risk analysis
analysis_df = analysis_df[
    ~problem_mask
].copy()

print("Original rows:", len(master_df))
print("Analysis rows:", len(analysis_df))
```

```python
volatility_summary_clean = (
    analysis_df
    .groupby("Index")["Daily_Return"]
    .agg(
        Mean_Daily_Return="mean",
        Daily_Volatility="std"
    )
    .reset_index()
)

volatility_summary_clean["Annualized_Volatility"] = (
    volatility_summary_clean["Daily_Volatility"]
    * (252 ** 0.5)
)

volatility_summary_clean = (
    volatility_summary_clean.round(4)
)

display(volatility_summary_clean)
```

```python
drawdown_results_clean = []

for index_name, data in analysis_df.groupby("Index"):

    prices = data["Close"]

    running_peak = prices.cummax()

    drawdown = (
        (prices - running_peak) /
        running_peak
    ) * 100

    max_drawdown = drawdown.min()

    drawdown_results_clean.append({
        "Index": index_name,
        "Maximum_Drawdown_%": max_drawdown
    })

drawdown_summary_clean = pd.DataFrame(
    drawdown_results_clean
)

drawdown_summary_clean[
    "Maximum_Drawdown_%"
] = drawdown_summary_clean[
    "Maximum_Drawdown_%"
].round(2)

display(drawdown_summary_clean)
```

```python
# CORRECTED MAXIMUM DRAWDOWN
drawdown_results = []

for index_name, data in analysis_df.groupby("Index"):

    data = data.sort_values("Date").copy()

    # Special handling for NIFTY IT:
    # Calculate drawdown using the consistent
    # post-discontinuity series.
    if index_name == "NIFTY IT":
        data = data[
            data["Date"] > pd.Timestamp("2004-05-28")
        ].copy()

    prices = data["Close"]

    running_peak = prices.cummax()

    drawdown = (
        (prices - running_peak) /
        running_peak
    ) * 100

    drawdown_results.append({
        "Index": index_name,
        "Maximum_Drawdown_%": drawdown.min()
    })

drawdown_summary_final = pd.DataFrame(drawdown_results)

drawdown_summary_final["Maximum_Drawdown_%"] = (
    drawdown_summary_final["Maximum_Drawdown_%"].round(2)
)

display(drawdown_summary_final)
```

```python
# FINAL RISK SUMMARY
risk_summary = volatility_summary_clean[
    ["Index", "Mean_Daily_Return", "Daily_Volatility",
     "Annualized_Volatility"]
].merge(
    drawdown_summary_final,
    on="Index"
)

risk_summary = risk_summary.round(4)

display(risk_summary)
```

```python
# STRONGEST CORRELATION PAIRS
import numpy as np

corr_pairs = []

indices = correlation_matrix.columns

for i in range(len(indices)):
    for j in range(i + 1, len(indices)):

        corr_value = correlation_matrix.iloc[i, j]

        corr_pairs.append({
            "Index_1": indices[i],
            "Index_2": indices[j],
            "Correlation": corr_value
        })

corr_pairs_df = pd.DataFrame(corr_pairs)

corr_pairs_df["Abs_Correlation"] = (
    corr_pairs_df["Correlation"].abs()
)

corr_pairs_df = corr_pairs_df.sort_values(
    "Abs_Correlation",
    ascending=False
)

display(
    corr_pairs_df[
        ["Index_1", "Index_2", "Correlation"]
    ].head(10).round(3)
)
```

**Statistical Analysis**

```python
# PREPARE DATA FOR ANOVA
return_groups = []

for index_name, data in analysis_df.groupby("Index"):

    returns = data["Daily_Return"].dropna()

    return_groups.append(returns)

print("Number of index groups:", len(return_groups))

for i, group in enumerate(return_groups):
    print(
        f"Group {i+1}: {len(group)} observations"
    )
```

```python
# ONE-WAY ANOVA
from scipy.stats import f_oneway

anova_stat, anova_p = f_oneway(
    *return_groups
)

print("One-Way ANOVA Results")
print("=====================")

print(f"F-statistic: {anova_stat:.4f}")
print(f"p-value: {anova_p:.6f}")

alpha = 0.05

if anova_p < alpha:
    print(
        "Result: Statistically significant difference "
        "detected among mean daily returns."
    )
else:
    print(
        "Result: No statistically significant difference "
        "detected among mean daily returns."
    )
```

```python
# ANOVA SUMMARY TABLE
anova_summary = pd.DataFrame({
    "Test": ["One-Way ANOVA"],
    "F_Statistic": [anova_stat],
    "P_Value": [anova_p],
    "Significance_Level": [alpha],
    "Conclusion": [
        "Significant difference"
        if anova_p < alpha
        else "No significant difference"
    ]
})

anova_summary = anova_summary.round(6)

display(anova_summary)
```

```python
# GOLD VS NIFTY T-TEST
from scipy.stats import ttest_ind

nifty_returns = (
    gold_nifty["NIFTY_50_Return"]
    .dropna()
)

gold_returns = (
    gold_nifty["Gold_Return"]
    .dropna()
)

t_stat, p_value = ttest_ind(
    nifty_returns,
    gold_returns,
    equal_var=False
)

print("Gold vs NIFTY 50 Independent t-test")
print("====================================")

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.6f}")

if p_value < 0.05:
    print(
        "Result: The mean daily returns show a "
        "statistically significant difference."
    )
else:
    print(
        "Result: No statistically significant difference "
        "was detected in mean daily returns."
    )
```
